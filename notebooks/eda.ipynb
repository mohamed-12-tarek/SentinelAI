```python
corr_pairs = (
    correlation
    .where(
        np.triu(
            np.ones(correlation.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .reset_index()
)

corr_pairs.columns = [
    "Feature 1",
    "Feature 2",
    "Correlation"
]

corr_pairs["Absolute Correlation"] = (
    corr_pairs["Correlation"].abs()
)

high_corr = corr_pairs[
    corr_pairs["Absolute Correlation"] >= 0.90
].sort_values(
    "Absolute Correlation",
    ascending=False
)

display(high_corr)
```

```python
suspicious_features = [
    "Fwd Header Length",
    "Bwd Header Length",
    "Total Length of Bwd Packet",
    "Total TCP Flow Time",
    "Flow Bytes/s",
    "Flow Packets/s"
]

review_report = pd.DataFrame(index=suspicious_features)

review_report["Min"] = df[suspicious_features].min()
review_report["Max"] = df[suspicious_features].max()

review_report["Negative Count"] = (
    (df[suspicious_features] < 0).sum()
)

review_report["Negative %"] = (
    review_report["Negative Count"] / len(df) * 100
)

review_report["Zero Count"] = (
    (df[suspicious_features] == 0).sum()
)

display(review_report)
```

```python
zero_pairs = [
    ("Flow Bytes/s", "Packet Length Mean"),
    ("Total Bwd packets", "Flow Bytes/s"),
    ("Total Bwd packets", "Packet Length Mean")
]

for col1, col2 in zero_pairs:
    z1 = df[col1] == 0
    z2 = df[col2] == 0

    print(f"\n{col1} vs {col2}")
    print("Both Zero:", (z1 & z2).sum())
    print("First Only:", (z1 & ~z2).sum())
    print("Second Only:", (~z1 & z2).sum())
    print("Neither Zero:", (~z1 & ~z2).sum())
```

```python
tcp_df = df[df["Protocol"] == 6].copy()

tcp_df["TCP_to_Flow_Ratio"] = (
    tcp_df["Total TCP Flow Time"] /
    tcp_df["Flow Duration"]
)

print("Ratio == 1:")
print(
    np.isclose(
        tcp_df["TCP_to_Flow_Ratio"],
        1,
        rtol=1e-5,
        atol=1e-8
    ).sum()
)

print("\nRatio > 1:")
print((tcp_df["TCP_to_Flow_Ratio"] > 1).sum())

print("\nRatio < 1:")
print((tcp_df["TCP_to_Flow_Ratio"] < 1).sum())

print("\nRatio Statistics:")
display(
    tcp_df["TCP_to_Flow_Ratio"].describe(
        percentiles=[0.5, 0.9, 0.95, 0.99, 0.999]
    )
)
```

```python
ratio_gt1 = tcp_df[
    tcp_df["TCP_to_Flow_Ratio"] > 1
].copy()

print("Rows:", len(ratio_gt1))

print("\nDistribution by Label:")
display(
    ratio_gt1["Label"].value_counts().to_frame("Count")
)

print("\nRatio Statistics:")
display(
    ratio_gt1["TCP_to_Flow_Ratio"].describe(
        percentiles=[0.5, 0.9, 0.95, 0.99, 0.999]
    )
)
```

```python
ratio_gt1 = tcp_df[
    tcp_df["TCP_to_Flow_Ratio"] > 1
].copy()

print("Rows:", len(ratio_gt1))

print("\nDistribution by Label:")
display(
    ratio_gt1["Label"].value_counts().to_frame("Count")
)

print("\nRatio Statistics:")
display(
    ratio_gt1["TCP_to_Flow_Ratio"].describe(
        percentiles=[0.5, 0.9, 0.95, 0.99, 0.999]
    )
)
```
